## Section 5.1: Feature Selection Using Correlation

Correlation-based feature selection was performed to identify the numerical variables with the strongest relationships with the target variable, `selling_price`.

The features were ranked according to their absolute correlation with `selling_price`. The results showed that `max_power` had the strongest correlation at 0.750, followed by `engine` at 0.586, `mileage` at 0.306, and `vehicle_age` at 0.242.

The four highest-scoring numerical features were therefore selected using K = 4:

- `max_power`
- `engine`
- `mileage`
- `vehicle_age`

The remaining numerical variables, `seats` and `km_driven`, had weaker correlations with the target, with absolute correlation values of 0.115 and 0.080 respectively. They were therefore not included among the four highest-scoring numerical features.

The selected features provide a focused set of numerical variables for further modelling analysis. However, correlation only measures linear relationships. Categorical variables such as brand, model, fuel type, transmission type, and seller type may also contain useful predictive information and will therefore be considered during the machine learning stage.

```python
# Define the target variable
y = df['selling_price']

# Define the selected numerical features
selected_numerical_features = [
    'max_power',
    'engine',
    'mileage',
    'vehicle_age'
]

# Define the categorical features
categorical_features = [
    'car_name',
    'brand',
    'model',
    'seller_type',
    'fuel_type',
    'transmission_type'
]

# Combine numerical and categorical features
feature_columns = selected_numerical_features + categorical_features

# Create the feature dataset
X = df[feature_columns]

# Display the shape of the features and target
print("Feature dataset shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Import the train-test split function
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Display the shapes of the resulting datasets
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Import preprocessing tools
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# Create the preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), selected_numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

# Fit the preprocessing steps using the training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the same preprocessing to the testing data
X_test_processed = preprocessor.transform(X_test)

# Display the resulting dimensions
print("Processed training data shape:", X_train_processed.shape)
print("Processed testing data shape:", X_test_processed.shape)
```

```python
# Import the Linear Regression algorithm
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
linear_model = LinearRegression()

# Train the model using the processed training data
linear_model.fit(X_train_processed, y_train)

# Make predictions on the test data
y_pred_linear = linear_model.predict(X_test_processed)

# Display a few predictions
y_pred_linear[:10]
```

```python
# Import regression evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Calculate evaluation metrics
mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))
r2_linear = r2_score(y_test, y_pred_linear)

# Display the results
print("Linear Regression Results")
print("-------------------------")
print(f"MAE:  {mae_linear:,.2f}")
print(f"RMSE: {rmse_linear:,.2f}")
print(f"R²:   {r2_linear:.4f}")
```

```python
# Import the Random Forest Regressor
from sklearn.ensemble import RandomForestRegressor

# Create the Random Forest model
random_forest_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train the model
random_forest_model.fit(X_train_processed, y_train)

# Make predictions on the test data
y_pred_rf = random_forest_model.predict(X_test_processed)

# Display a few predictions
y_pred_rf[:10]
```

```python
# Calculate evaluation metrics for the Random Forest model
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

# Display the results
print("Random Forest Results")
print("---------------------")
print(f"MAE:  {mae_rf:,.2f}")
print(f"RMSE: {rmse_rf:,.2f}")
print(f"R²:   {r2_rf:.4f}")
```

```python
# Import the Gradient Boosting Regressor
from sklearn.ensemble import GradientBoostingRegressor

# Create the Gradient Boosting model
gradient_boosting_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

# Train the model
gradient_boosting_model.fit(X_train_processed, y_train)

# Make predictions on the test data
y_pred_gb = gradient_boosting_model.predict(X_test_processed)

# Display a few predictions
y_pred_gb[:10]
```

```python
# Calculate evaluation metrics for the Gradient Boosting model
mae_gb = mean_absolute_error(y_test, y_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))
r2_gb = r2_score(y_test, y_pred_gb)

# Display the results
print("Gradient Boosting Results")
print("-------------------------")
print(f"MAE:  {mae_gb:,.2f}")
print(f"RMSE: {rmse_gb:,.2f}")
print(f"R²:   {r2_gb:.4f}")
```

```python
# Create a comparison table for the three baseline models
model_results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Random Forest',
        'Gradient Boosting'
    ],
    'MAE': [
        mae_linear,
        mae_rf,
        mae_gb
    ],
    'RMSE': [
        rmse_linear,
        rmse_rf,
        rmse_gb
    ],
    'R2': [
        r2_linear,
        r2_rf,
        r2_gb
    ]
})

# Display the results
model_results
```

## Section 6.1: Baseline Model Comparison

Three regression algorithms were trained and evaluated using the same training and testing datasets: Linear Regression, Random Forest Regressor, and Gradient Boosting Regressor.

The Linear Regression model achieved an MAE of 178,516.10, an RMSE of 385,236.04, and an R² of 0.8029. This provided a baseline for comparing the more complex models.

The Random Forest model achieved an MAE of 99,476.00, an RMSE of 216,498.65, and an R² of 0.9377. Among the three baseline models, it produced the lowest MAE and RMSE and the highest R².

The Gradient Boosting model achieved an MAE of 129,971.99, an RMSE of 261,873.54, and an R² of 0.9089. Its performance was better than Linear Regression but lower than the Random Forest model based on the current test-set results.

These results indicate that the tree-based ensemble models captured relationships in the data that were not fully captured by the linear model. However, the models will undergo hyperparameter tuning before the final model is selected.

```python
# Import GridSearchCV for hyperparameter tuning
from sklearn.model_selection import GridSearchCV

# Create a Random Forest model for tuning
rf_for_tuning = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

# Define the hyperparameter combinations to test
rf_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

# Create the GridSearchCV object
rf_grid_search = GridSearchCV(
    estimator=rf_for_tuning,
    param_grid=rf_param_grid,
    cv=3,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1,
    verbose=1
)

# Run the hyperparameter search
rf_grid_search.fit(X_train_processed, y_train)

# Display the best parameters
print("Best Random Forest parameters:")
print(rf_grid_search.best_params_)

# Display the best cross-validation RMSE
print("\nBest cross-validation RMSE:")
print(-rf_grid_search.best_score_)
```
